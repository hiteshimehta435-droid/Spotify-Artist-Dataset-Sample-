Explore visualization

Bar chart

```python
import pandas as pd
import matplotlib.pyplot as plt

Spotify = pd.read_csv(r"F:\New folder (6)\python note\python wrok\Most Streamed Artists on Spotify (17_07_2026).csv")

plt.bar(Spotify["Sex"], Spotify["Total Streams (in millions)"],color='purple')
plt.title("Monthly Listeners")
plt.xlabel("Artist")
plt.ylabel("Total Streams (in millions)")
plt.xticks(rotation=45)
plt.show()
```

Pie chart

```python
import pandas as pd
import matplotlib.pyplot as plt

Spotify = pd.read_csv(r"F:\New folder (6)\python note\python wrok\Most Streamed Artists on Spotify (17_07_2026).csv")

top = Spotify.head(5)

plt.pie(
    top["Total Streams (in millions)"],
    labels=top["Langugage"],
    autopct="%1.1f%%"
)

plt.title("Top 5 Monthly Listeners")
plt.show()
```

Scatter chart

```python
import pandas as pd
import matplotlib.pyplot as plt

Spotify = pd.read_csv(r"F:\New folder (6)\python note\python wrok\Most Streamed Artists on Spotify (17_07_2026).csv")

plt.scatter(
    Spotify["Total Streams (in millions)"],
    Spotify["Solo Streams (in millions)"]
)

plt.xlabel("Total Streams (in millions)")
plt.ylabel("Solo Streams (in millions)")
plt.title("Total Streams (in millions) vs Solo Streams (in millions)")
plt.show()
```
